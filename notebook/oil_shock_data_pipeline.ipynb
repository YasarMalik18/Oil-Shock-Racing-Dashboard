```python
import pandas as pd
import numpy as np
```

```python
dates = pd.date_range(start="2018-01-01", end="2024-12-01", freq='MS')

oil_price = np.cumsum(np.random.normal(0, 2, len(dates))) + 70

shock = np.random.choice(["Normal", "Demand Shock", "Supply Shock"], len(dates))
shock_mag = np.random.randint(-60, 60, len(dates))

oil_df = pd.DataFrame({
    "Date": dates,
    "Oil_Price": oil_price.round(2),
    "Shock_Type": shock,
    "Shock_Magnitude": shock_mag
})

oil_df.to_csv("../data/oil_prices.csv", index=False)
```

```python
sectors = ["Energy", "Airlines", "Transport", "Manufacturing"]

sector_df = pd.DataFrame({
    "Date": np.repeat(dates, len(sectors)),
    "Sector": sectors * len(dates),
    "Return": np.random.normal(5, 12, len(dates)*len(sectors)).round(2)
})

sector_df.to_csv("../data/sector_data.csv", index=False)
```

```python
macro_df = pd.DataFrame({
    "Date": dates,
    "Inflation": np.random.uniform(2, 8, len(dates)).round(2),
    "GDP": np.random.uniform(-3, 5, len(dates)).round(2),
    "Interest_Rate": np.random.uniform(3, 8, len(dates)).round(2)
})

macro_df.to_csv("../data/macro_data.csv", index=False)
```

```python
import pandas as pd
```

```python
oil_df = pd.read_csv("../data/oil_prices.csv")
sector_df = pd.read_csv("../data/sector_data.csv")
macro_df = pd.read_csv("../data/macro_data.csv")
```

```python
oil_df['Date'] = pd.to_datetime(oil_df['Date'])
sector_df['Date'] = pd.to_datetime(sector_df['Date'])
macro_df['Date'] = pd.to_datetime(macro_df['Date'])
```

```python
oil_df.dropna(inplace=True)
sector_df.dropna(inplace=True)
macro_df.dropna(inplace=True)

oil_df.drop_duplicates(inplace=True)
sector_df.drop_duplicates(inplace=True)
macro_df.drop_duplicates(inplace=True)
```

```python
oil_df['Oil_Return'] = oil_df['Oil_Price'].pct_change()*100

oil_df['Shock_Level'] = oil_df['Shock_Magnitude'].apply(
    lambda x: "High" if abs(x)>40 else "Medium" if abs(x)>20 else "Low"
)

oil_df['Year'] = oil_df['Date'].dt.year
oil_df['Month'] = oil_df['Date'].dt.month
```

```python
df = oil_df.merge(sector_df, on="Date").merge(macro_df, on="Date")
```

```python
df.to_csv("cleaned_data.csv", index=False)
df.to_csv("final_dashboard_data.csv", index=False)
```
